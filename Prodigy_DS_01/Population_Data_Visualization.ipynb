```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    'API_SP.POP.TOTL_DS2_en_csv_v2_127039.csv',
    skiprows=4
)
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
population_df = df[['Country Name', '2022']]
```

```python
population_df.columns = [
    'Country',
    'Population'
]
```

```python
population_df.dropna(inplace=True)
```

```python
top10 = population_df.sort_values(
    by='Population',
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x='Population',
    y='Country',
    data=top10
)

plt.title('Top 10 Most Populated Countries')

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    population_df['Population'],
    bins=30,
    kde=True
)

plt.title('Population Distribution')

plt.xlabel('Population')

plt.show()
```

```python
plt.figure(figsize=(10,10))

plt.pie(
    top10['Population'],
    labels=top10['Country'],
    autopct='%1.1f%%'
)

plt.title('Population Share of Top 10 Countries')

plt.show()
```

```python
countries = df[
    df['Country Name'].isin(
        ['India', 'China', 'United States']
    )
]

years = ['2000', '2005', '2010', '2015', '2020', '2022']

countries.set_index(
    'Country Name'
)[years].T.plot(
    figsize=(12,6)
)

plt.title('Population Growth Comparison')

plt.xlabel('Year')

plt.ylabel('Population')

plt.show()
```

```python
countries = df[
    df['Country Name'].isin(
        ['India', 'China', 'United States']
    )
]

years = ['2000', '2005', '2010', '2015', '2020', '2022']

countries.set_index(
    'Country Name'
)[years].T.plot(
    figsize=(12,6)
)

plt.title('Population Growth Comparison')

plt.xlabel('Year')

plt.ylabel('Population')

plt.show()
```

```python
lowest = population_df.sort_values(
    by='Population'
).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x='Population',
    y='Country',
    data=lowest
)

plt.title('Lowest Population Countries')

plt.show()
```
