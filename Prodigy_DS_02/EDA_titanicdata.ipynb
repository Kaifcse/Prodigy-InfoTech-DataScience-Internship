```python
df['individual_fare'] = df['Fare']/(df['SibSp'] + df['Parch'] + 1)
```

```python
df[['individual_fare','Fare']].describe()
```

```python
df['family_size'] = df['SibSp'] + df['Parch'] + 1
```

```python
df['family_size']
```

```python
def transfer_family_size(num):
  if num ==1:
    return 'alone'
  elif num>1 and num<5:
      return "small"
  else:
        return "large"
```

```python
df[['individual_fare','Fare']]
```

```python
df['family_type']=df['family_size'].apply(transfer_family_size) #we make acolumn and merge
```

```python
df['family_type']
```

```python
pd.crosstab(df['Survived'],df['family_type'],normalize='columns')*100
```
